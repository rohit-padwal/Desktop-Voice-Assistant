```python
!pip install pyaudio
!pip install SpeechRecognition
!pip install gtts
!pip install playsound
!pip install pywhatkit
!pip install wikipedia
!pip install pyjokes
!sudo apt-get install espeak
!pip install pyttsx3
!apt install libasound2-dev portaudio19-dev libportaudio2 libportaudiocpp0 ffmpeg

import speech_recognition as sr
import pyttsx3
# import pywhatkit
import datetime
import wikipedia
import pyjokes
import os
import webbrowser
import smtplib
import requests
import yfinance as yf
from newsapi import NewsApiClient

listener = sr.Recognizer()
engine = pyttsx3.init()
voices = engine.getProperty('voices')
engine.setProperty('voice', voices[1].id)

def talk(text):
    engine.say(text)
    engine.runAndWait()

def take_command():
    command = ''

    try:
        with sr.Microphone() as source:
            print('listening...')
            voice = listener.listen(source)
            command = listener.recognize_google(voice)
            command = command.lower()
            if 'assistant' in command:
                command = command.replace('assistant', '')
                print(command)
    except:
        pass
    return command

def run_assistant():
    command = take_command()
    if 'play' in command:
        song = command.replace('play', '')
        talk('playing ' + song)
        url = 'https://www.youtube.com/results?search_query=' + song
        webbrowser.get().open(url)
    elif 'search' in command:
        search = command.replace('search', '')
        talk('searching ' + search)
        url = 'https://www.google.com/search?q=' + search
        webbrowser.get().open(url)
    elif 'time' in command:
        time = datetime.datetime.now().strftime('%I:%M %p')
        talk('the time is ' + time)
    elif 'date' in command:
        date = datetime.datetime.now().strftime('%A %d %B %Y')
        talk('today is ' + date)
    elif 'tell me about' in command:
        search_term = command.replace('tell me about', '')
        info = wikipedia.summary(search_term, 1)
        talk(info)
    elif 'joke' in command:
        talk(pyjokes.get_joke())
    elif 'who is' in command:
        person = command.replace('who is', '')
        info = wikipedia.summary(person, 1)
        print(info)
        talk(info)
    elif 'send email to' in command:
        try:
            to = command.split('send email to')[1].split(' ')[1]
            talk('What should I say?')
            body = take_command()
            subject = 'Email from Python Assistant'
            send_email(to, subject, body)
        except:
            talk('Sorry, I could not send the email.')
    elif 'weather in' in command:
        city = command.replace('weather in', '')
        get_weather(city)
    elif 'stock price of' in command:
        ticker = command.split(' ')[-1]
        get_stock_price(ticker)
    elif 'news headlines' in command:
        get_news_headlines()
    elif 'stop' in command:
        talk('Goodbye!')
        return False
    else:
        talk('Sorry, I did not understand that.')
    return True

def search_wikipedia(query):
    try:
        result = wikipedia.summary(query, sentences=2)
        talk(result)
    except:
        talk("Sorry, I couldn't find anything on Wikipedia for that.")
        
def send_email(to, subject, body):
    # Add your email credentials here
    email = 'your_email_address'
    password = 'your_email_password'
    
    server = smtplib.SMTP('smtp.gmail.com', 587)
    server.starttls()
    server.login(email, password)
    
    message = f'Subject: {subject}\n\n{body}'
    server.sendmail(email, to, message)
    
    talk('Email sent successfully.')
    
def get_weather(city):
    # Add your OpenWeatherMap API key here 
    api_key = 'bd5e378503939ddaee76f12ad7a97608'
    
    url = f'http://api.openweathermap.org/data/2.5/weather?q={city}&appid={api_key}'
    response = requests.get(url)
    data = response.json()
    
    temp = data['main']['temp']
    feels_like = data['main']['feels_like']
    description = data['weather'][0]['description']
    
    talk(f'The temperature in {city} is {temp} degrees Celsius. It feels like {feels_like} degrees.')
    talk(f'Outside, it is {description}.')

def get_stock_price(ticker):
    stock = yf.Ticker(ticker)
    current_price = stock.info['regularMarketPrice']
    
    talk(f'The current stock price of {ticker} is {current_price}.')

def get_news_headlines():
    # Add your NewsAPI API key here
    api_key = 'your_api_key'
    
    newsapi = NewsApiClient(api_key)
    top_headlines = newsapi.get_top_headlines(language='en')
    
    talk('Here are some of the top headlines from the news today.')
    for article in top_headlines['articles'][:5]:
        talk(article['title'])

talk('Hi, how can I assist you?')
while True:
    if not run_assistant():
        break

```
